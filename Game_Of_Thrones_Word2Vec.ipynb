```python

import nltk
from nltk import sent_tokenize
from gensim.utils import simple_preprocess

story=[]
```

```python
for filename in os.listdir('/content/drive/MyDrive/GameOfThrones'):

    file_path = os.path.join('/content/drive/MyDrive/GameOfThrones', filename)

    with open(file_path, encoding='cp1252') as f:
        corpus = f.read()

    raw_sent = sent_tokenize(corpus)

    for sent in raw_sent:
        story.append(simple_preprocess(sent))

print(len(story))
```

```python
story
```

```python
from nltk.corpus import stopwords

def stop_words(story):
  stop_word = set(stopwords.words('english'))
  cleaned_story = []
  for sentence in story:
    cleaned_sent = []
    for word in sentence:
      if word not in stop_word:
        cleaned_sent.append(word)
    cleaned_story.append(cleaned_sent)
  return cleaned_story

cleaned_story = stop_words(story)

print(story[0])
print(cleaned_story[0])
```

```python
model = gensim.models.Word2Vec(window=10, min_count=2)
model.build_vocab(story)
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
model.wv.most_similar('sign')
```

```python
model.wv.doesnt_match(['jon','rikon','rob','arya','brown'])
model.wv['jon']
```

```python
model.wv.similarity('arya','count')
```
